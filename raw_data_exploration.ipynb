```python
diseases = pd.read_parquet("data/raw/disease/")
targets = pd.read_parquet("data/raw/target/")
associations = pd.read_parquet("data/raw/association_overall_direct/")
```

```python
print("Diseases: ", diseases.shape, "->", diseases.columns.to_list()[:5])
print("Targets: ", targets.shape, "->", targets.columns.to_list()[:5])
print("Associations: ", associations.shape, "->", associations.columns.to_list()[:5])
```

```python
alzheimer_id = diseases[diseases['name'] == 'Alzheimer disease']['id'].iloc[0]
alz_associations = associations[associations['diseaseId'] == alzheimer_id]

```

```python
alz_targets = targets.merge(alz_associations, left_on='id', right_on='targetId')
```

```python
alz_targets.sort_values(by = ['score'],ascending = False)[:10]
```
